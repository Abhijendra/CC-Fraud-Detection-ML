```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
credit_card_data = pd.read_csv("/content/creditcard.csv")
```

# Preprocessing

```python
credit_card_data.head(5)
```

```python
credit_card_data.info()
```

```python
credit_card_data.isnull().sum()
```

```python
credit_card_data['Class'].value_counts()
```

```python
legit_data = credit_card_data[credit_card_data.Class == 0]
fraud_data = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit_data.shape)
print(fraud_data.shape)
```

```python
# stastical measures
legit_data.Amount.describe()
```

```python
fraud_data.Amount.describe()
```

```python
# compare the values for both the transactions

credit_card_data.groupby('Class').mean()
```

# Undersampling

Build a sample dataset containing similar distribution of normal transaction and fraud transactions

```python
legit_sample = legit_data.sample(n=492)

# Concatenating both DFs

new_dataset = pd.concat([legit_sample, fraud_data], axis=0)
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

```python
# Splitting data into Features and targets

X = new_dataset.drop('Class', axis=1)
Y = new_dataset['Class']
```

```python
print(X)
```

```python
print(Y)
```

# Split the data into train and test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

# Model training

```python
# Logistic Regression Model

model = LogisticRegression()
```

```python
# training the model with training data

model.fit(X_train, Y_train)
```

# Evaluation of Model

### Accuracy on train data

```python
Y_train_predicted = model.predict(X_train)

training_data_accuracy = accuracy_score(Y_train_predicted, Y_train)

print(training_data_accuracy)
```

### Accuracy on test data


```python

Y_test_predicted = model.predict(X_test)

test_data_accuracy = accuracy_score(Y_test_predicted, Y_test)

print(test_data_accuracy)
```
